```python
import openeo
from openeo.processes import quantiles

import xarray as xr
import os
```

```python
connection = openeo.connect(url="openeo.dataspace.copernicus.eu")
connection.authenticate_oidc()
```

## Sentinel-1 mosaic from Sentinel-1 GRD

Product description: 

https://documentation.dataspace.copernicus.eu/Data/SentinelMissions/Sentinel1.html#sentinel-1-level-3-monthly-mosaics

```python
spatial_extent_west = {"west": 610950, "east": 674650, "south": 5143820, "north": 5206380, "crs": "EPSG:32632"} # Western South Tyrol glaciers
#spatial_extent_east = {"west": 704100, "east": 747040, "south": 5195880, "north": 5219660, "crs": "EPSG:32632"} # Eastern South Tyrol glaciers
temporal_extent = ["2017-07-01T00:00:00Z", "2017-09-30T23:59:59Z"]

output_folder = "/mnt/CEPH_PROJECTS/provinzBZ_risk_EO/S1_mosaics/"
```

```python
sentinel1 = connection.load_collection(
    "SENTINEL1_GLOBAL_MOSAICS",
    temporal_extent = temporal_extent,
    spatial_extent = spatial_extent_west,
    bands = ["VV","VH"],
)
```

```python
job = sentinel1.create_job(title='S1_mosaic')
job.start_and_wait()
```

```python
job.get_results().download_files(output_folder)
```
